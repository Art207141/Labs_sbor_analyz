# 7. Группировка и получение аналитического результата

```python
# 7.1
store_df["is_high_value"] = store_df["order_sum"] > 3000
print(store_df[["order_id", "order_sum", "is_high_value"]])
```

## Этап 7.1 — Создание производного признака

С помощью конструкции `store_df["order_sum"] > 3000` в структуру таблицы внедрен новый бинарный признак `is_high_value`. Маркер принимает значение `True`, если затраты на заказ превысили планку в 3000 рублей, и `False` во всех остальных случаях.

Под критерий дорогих покупок подпали 4 транзакции (идентификаторы: 1004, 1008, 1011, 1015).


```python
# 7.2
channel_stats = store_df.groupby("channel")["order_sum"].mean()
print(channel_stats)
```

## Этап 7.2 — Группировка по каналу продаж

Операция `groupby("channel")["order_sum"].mean()` позволила рассчитать средние чеки для каждого источника продаж:
- Мобильное приложение (app): 1800.00
- Внешние площадки (marketplace): 13775.00
- Веб-сайт (site): 1816.67

Финансовые показатели маркетплейсов превосходят альтернативные каналы почти в 7 раз. Данный феномен вызван исключительно присутствием аномальной транзакции на 45 000 рублей в этой группе. Без учета данного фактора средний чек на маркетплейсах находился бы в районе 3 367 рублей.


```python
# 7.3
channel_stats = channel_stats.sort_values(ascending=False)
print(channel_stats)
```

## Этап 7.3 — Сортировка результата

Применив метод `sort_values(ascending=False)`, мы упорядочили торговые каналы по весу их среднего чека от большего к меньшему:
1. Площадки marketplace — 13 775.00
2. Официальный веб-ресурс site — 1 816.67
3. Мобильный клиент app — 1 800.00


## Этап 7.4 — Комплексная аналитическая интерпретация

1. Средние чеки по источникам поступления: marketplace лидирует с 13775.00, тогда как site (1816.67) и app (1800.00) демонстрируют схожие скромные результаты.
2. Объем маржинальных операций: зафиксировано 4 крупных чека с флагом `is_high_value = True`.
3. Состояние информации: выявлено 3 пробела в логах времени обслуживания клиентов.
4. Специфика аномалий: единичный заказ на 45000 рублей искажает метрику среднего чека для маркетплейсов. В результате общее среднее по компании (5000) оторвано от медианы (2100).
5. Параметры времени: показатели среднего (39.58) и медианы (37.5) для логистики практически совпадают, демонстрируя отсутствие критических сбоев.

**Резюме:**  
Финансовая структура продаж сильно деформирована из-за одной сверхкрупной сделки на маркетплейсе, которая маскирует реальную ситуацию в каналах site и app. При аудите чеков правильнее ориентироваться на медиану (2100). В то же время показатели скорости доставки выглядят монолитными и предсказуемыми. Около трети базы составляют крупные заказы, а 20% логов имеют пробелы в фиксации времени.
